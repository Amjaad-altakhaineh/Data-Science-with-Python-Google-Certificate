## Diabetes Dataset Description

You are working as a Biomedical Engineering researcher and have been given a dataset called `diabetes.csv`.

The goal is to predict whether a patient has diabetes (`Outcome` column) from several medical measurements.

All patients are female and at least 21 years old.

**Columns:**

- `Pregnancies`: number of pregnancies  
- `Glucose`: plasma glucose concentration  
- `BloodPressure`: diastolic blood pressure (mm Hg)  
- `SkinThickness`: triceps skin fold thickness (mm)  
- `Insulin`: 2-hour serum insulin (µU/ml)  
- `BMI`: body mass index (weight / height²)  
- `DiabetesPedigreeFunction`: score based on family history  
- `Age`: age in years  
- `Outcome`: 1 = diabetes, 0 = no diabetes

## Q15 – Covariance of `Age` and `Glucose`

```python
# Q15
cov_age_glucose = data[["Age", "Glucose"]].cov()
cov_age_glucose
```

## Q16 – Meaning of the covariance matrix (sample answer)

The table from Q15 has two rows and two columns:

- Top left: variance of `Age`  
- Bottom right: variance of `Glucose`  
- Other two values: covariance between `Age` and `Glucose`

The diagonal values show how much each variable spreads around its mean.  
The other values show how `Age` and `Glucose` move together.

## Q17 – Relationship between `Age` and `Glucose` (sample answer)

Look at the covariance value between `Age` and `Glucose`:

- If it is **positive and not tiny**:  
  When `Age` is larger, `Glucose` tends to be larger.  

- If it is **negative**:  
  When `Age` is larger, `Glucose` tends to be smaller.  

- If it is **close to zero**:  
  There is no strong linear relationship.

In this dataset, the covariance is usually **positive**, so older patients
tend to have slightly higher glucose values.

## Q20 – Correlation matrix

```python
# Q20
corr_matrix = data.corr()
corr_matrix
```

## Q21 – Correlation with `Outcome` and correlation heatmap

```python
# Q21 – correlations with Outcome
print(corr_matrix["Outcome"])

# Heatmap of the full correlation matrix
sns.heatmap(corr_matrix, annot=True)
plt.title("Correlation matrix")
plt.show()
```

### Interpreting correlations with `Outcome` (sample answer)

Values are between `-1` and `1`:

- Close to `1` → strong positive relationship with diabetes  
- Close to `-1` → strong negative relationship  
- Close to `0` → weak or no linear relationship

Common pattern in this dataset:

- `Glucose` has one of the **strongest positive correlations** with `Outcome`.  
- `BMI`, `Age`, `DiabetesPedigreeFunction` and `Pregnancies` often have
  **moderate positive correlations**.  
- `BloodPressure`, `SkinThickness` and `Insulin` often have **small correlations**.

So `Glucose`, `BMI` and `Age` are more related to diabetes in this table.
